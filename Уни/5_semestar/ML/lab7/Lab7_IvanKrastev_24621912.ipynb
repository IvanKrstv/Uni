```python
import random
import nltk
from nltk.corpus import names

# Подготовка на данните
labeled_names = ([(name, 'male') for name in names.words('male.txt')] +
                [(name, 'female') for name in names.words('female.txt')])

random.seed(42)
random.shuffle(labeled_names)

train_names = labeled_names[:5000]
test_names = labeled_names[5000:]

# Метод с правила
def rule_based_gender(name):
  last_letter = name[-1].lower()
  gender = None

  if last_letter in ('a', 'e', 'i'):
    gender = 'female'
  else:
    gender = 'male'

  return gender


correct = 0
for name, true_gender in test_names:
  predicted_gender = rule_based_gender(name)
  if predicted_gender == true_gender:
    correct += 1

rule_accuracy = (correct / len(test_names)) * 100
print("Точност на метода с правила:", round(rule_accuracy, 2), "%")

# Наивен Бейсов класификатор
def gender_features(word):
  return {'last_letter': word[-1].lower()}


featuresets = [(gender_features(name), gender) for name, gender in labeled_names]

train_set, test_set = featuresets[:5000], featuresets[5000:]

classifier = nltk.NaiveBayesClassifier.train(train_set)
nb_train_accuracy = nltk.classify.accuracy(classifier, train_set) * 100
print("Точност на Naive Bayes модела при обучаващо множество:", round(nb_train_accuracy, 2), "%")
nb_accuracy = nltk.classify.accuracy(classifier, test_set) * 100
print("Точност на Naive Bayes модела при тестово множество:", round(nb_accuracy, 2), "%")

names_for_test = ['Neo', 'Trinity', 'Maria', 'John', 'Alex', 'Kim']
for name in names_for_test:
  print()
  print(f"Rule based    -> {name}: {rule_based_gender(name)}")
  print(f"Model predict -> {name}: {classifier.classify(gender_features(name))}")
  if name in names_for_test[:3]:
    dist = classifier.prob_classify(gender_features(name))
    print(f"Probability for a female: {round(dist.prob('female'), 3)}; Probability for a male: {round(dist.prob('male'), 3)}")
print()
classifier.show_most_informative_features(10)
```

```python
import random
import nltk
from nltk.corpus import names

# Подготовка на данните
labeled_names = ([(name, 'male') for name in names.words('male.txt')] +
                [(name, 'female') for name in names.words('female.txt')])

random.seed(42)
random.shuffle(labeled_names)

train_names = labeled_names[:5000]
test_names = labeled_names[5000:]

# Метод с правила
def rule_based_gender(name):
  last_letter = name[-1].lower()
  gender = None

  if last_letter in ('a', 'e', 'i'):
    gender = 'female'
  else:
    gender = 'male'

  return gender


correct = 0
for name, true_gender in test_names:
  predicted_gender = rule_based_gender(name)
  if predicted_gender == true_gender:
    correct += 1

rule_accuracy = (correct / len(test_names)) * 100
print("Точност на метода с правила:", round(rule_accuracy, 2), "%")

# Наивен Бейсов класификатор
def gender_features(word):
  return {'last_letter': word[-1].lower()}


def gender_features_v2(word):
  count_vowels = sum(1 for c in word.lower() if c in 'aeiouy')

  return {
      'last_letter': word[-1].lower(),
      'last_two_letters': word[-2:].lower(),
      'last_three_letters': word[-3:].lower(),
      'first_letter': word[0].lower(),
      'name_length': len(word),
      'count_vowels': count_vowels,
      'last_is_vowel': word[-1].lower() in 'aeiouy'
  }


featuresets = [(gender_features_v2(name), gender) for name, gender in labeled_names]

train_set, test_set = featuresets[:5000], featuresets[5000:]

classifier = nltk.NaiveBayesClassifier.train(train_set)
nb_train_accuracy = nltk.classify.accuracy(classifier, train_set) * 100
print("Точност на Naive Bayes модела при обучаващо множество:", round(nb_train_accuracy, 2), "%")
nb_accuracy = nltk.classify.accuracy(classifier, test_set) * 100
print("Точност на Naive Bayes модела при тестово множество:", round(nb_accuracy, 2), "%")

classifier.show_most_informative_features(10)
```

Задача 2

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from nltk.corpus import wordnet
from sklearn.naive_bayes import MultinomialNB

corpus = [
    "This movie was an absolute masterpiece with a fantastic story.",
    "A terrible and boring film, complete waste of time.",
    "An incredible visual experience, highly recommended to everyone.",
    "The plot was slow, predictable and frankly quite disappointing.",
    "The movie could not have been so fantastic and such masterpiece."
]
labels = ['pos', 'neg', 'pos', 'neg', 'pos']

# Bag-of-Words
bow_vectorizer = CountVectorizer(
    lowercase=True,
    stop_words='english',
    ngram_range=(1,2)
)

bow_matrix = bow_vectorizer.fit_transform(corpus)

print("Признаци:", bow_vectorizer.get_feature_names_out())
print("Размерност:", bow_matrix.shape)
print("Честотна матрица:\n", bow_matrix.toarray())

# TF-IDF
tfidf_vectorizer = TfidfVectorizer(
    stop_words='english'
)

tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)

print("Тегла в първия документ:")
terms = tfidf_vectorizer.get_feature_names_out()

for word, score in zip(terms, tfidf_matrix.toarray()[0]):
  if score > 0:
    print(f"{word} -> {score}")


# Разширяване на данните
def get_synonyms(word):
  synonyms = set()
  for syn in wordnet.synsets(word):
    for lemma in syn.lemmas():
      name = lemma.name().replace('_', ' ')
      if name.lower() != word.lower():
        synonyms.add(name)

  return sorted(synonyms)


for word in ['movie', 'good', 'slow', 'bank']:
  print(f"{word} -> {', '.join(get_synonyms(word))}")


def augment_text(sentence):
  words = sentence.split()
  variants = []
  for i in range(len(words)):
    clean_word = words[i].strip('.,!?')
    syns = get_synonyms(clean_word)
    if syns:
      new_words = words.copy()
      new_words[i] = words[i].replace(clean_word, syns[0])
      variants.append(' '.join(new_words))

  return variants


print("Изходно изречение:", corpus[0])
for idx, variant in enumerate(augment_text(corpus[0])[:3], 1):
  print("Вариант", idx, ":", variant)


aug_corpus, aug_labels = [], []
for text, label in zip(corpus, labels):
    aug_corpus.append(text)
    aug_labels.append(label)
    for variant in augment_text(text):
        aug_corpus.append(variant)
        aug_labels.append(label)

print("Брой примери: преди", len(corpus), "-> след", len(aug_corpus))

clf_vectorizer = TfidfVectorizer(stop_words='english')
X_train = clf_vectorizer.fit_transform(aug_corpus)

model = MultinomialNB()
model.fit(X_train, aug_labels)

new_reviews = [
    "A fantastic film with an incredible story, I highly recommend it.",
    "Boring and predictable plot, a complete waste of my time."
]
X_new = clf_vectorizer.transform(new_reviews)

for review, pred in zip(new_reviews, model.predict(X_new)):
    print(f"{review}\n  -> {pred}")
```

Разширяването на данните със синоними от WordNet увеличава броя на примерите, но не отчита контекста. Многозначни думи и съкращения се заменят с неподходящи синоними (например „was“ става „Evergreen State“, а „a“ става „adenine“), а някои изречения стават граматически грешни.
